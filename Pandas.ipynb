```python
import re

# Memilih teks spesifik
df[df['nama'].str.contains('Andi')]
df[df['nama'].str.contains('Jul')]

# Mengabaikan Huruf Besar/Kecil (Case Insensitive)
df[df['nama'].str.contains('in')]
df[df['nama'].str.contains('in', flags=re.IGNORECASE)]

# Memilih Data yang Berada di List/Dataframe Lain (.isin)
feb_penjualan = [130, 220]
df[df['feb'].isin(feb_penjualan)]
```

```python
import re
import pandas as pd

# 1. Sama dengan
df[df['jan'] == 200]
df[df.jan.eq(200)]

# 2. Tidak sama dengan
df[df['jan'] != 200]
df[df.jan.ne(200)]
df[~(df['jan'] == 200)]

# 3. Lebih dari
df[df['jan'] > 200]
df[df.jan.gt(200)]

# 4. Lebih dari atau Sama Dengan
df[df['jan'] >= 200]
df[df.jan.ge(200)]

# 5. Kurang dari
df[df['jan'] < 200]
df[df.jan.lt(200)]

# 6. Kurang dari atau Sama Dengan
df[df['jan'] <= 200]
df[df.jan.le(200)]
```
